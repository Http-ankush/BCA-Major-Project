```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Bookings.csv")
```

```python
print(df)
```

```python
print(df.columns.tolist())
```

```python
successful_bookings = df[df['Booking_Status'] == 'Success']
successful_bookings.head()
```

1. Retrieve all successful bookings

```python
customer_cancelled = df[df['Booking_Status'] == 'Canceled by Customer']

print("Total Cancelled Rides:", customer_cancelled.shape[0])
```

2. Find the average ride distance for each vehicle type

```python
df.groupby('Vehicle_Type')['Ride_Distance'].mean().reset_index()
```

4. List the top 5 customers who booked the highest number of rides

```python
top_5_customers = (
    df.groupby('Customer_ID')
      .size()
      .reset_index(name='Total_Rides')
      .sort_values(by='Total_Rides', ascending=False)
      .head(5)
)

top_5_customers
```

5. Number of rides cancelled by drivers due to personal and car-related issues

```python
driver_cancelled = (
    df['Canceled_Rides_by_Driver']
    == 'Personal & Car related issue'
).sum()

print("Total Rides:", driver_cancelled)
```

6. Find the maximum and minimum driver ratings for Prime Sedan bookings

```python
prime_sedan = df[df['Vehicle_Type'] == 'Prime Sedan']

max_rating = prime_sedan['Driver_Ratings'].max()
min_rating = prime_sedan['Driver_Ratings'].min()

print("Maximum Driver Rating:", max_rating)
print("Minimum Driver Rating:", min_rating)
```

7. Retrieve all rides where payment was made using UPI

```python
upi_rides = df[df['Payment_Method'] == 'UPI']

upi_rides.head()
```

8. Find the average customer rating per vehicle type

```python
avg_customer_rating = (
    df.groupby('Vehicle_Type')['Customer_Rating']
      .mean()
      .reset_index()
)

avg_customer_rating
```

9. Calculate the total booking value of rides completed successfully

```python
total_booking_value = (
    df[df['Booking_Status'] == 'Success']
    ['Booking_Value']
    .sum()
)

print("Total Booking Value:", total_booking_value)
```

10. List all incomplete rides along with the reason

```python
incomplete_rides = df[
    df['Incomplete_Rides'] == 'Yes'
][['Booking_ID',
   'Incomplete_Rides',
   'Incomplete_Rides_Reason']]

incomplete_rides
```
